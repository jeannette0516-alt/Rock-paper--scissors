```python
# ==========================================
# Python 輸出練習：請在「你的程式碼」下方完成作答
# ==========================================

# 1. 輸出文字 (此題為範例，不需修改)
print("歡迎來到 Python！")
print("-" * 30)

# 2. 多行字串輸出
# 提示：使用三個雙引號 """ """
# ✏️ 你的程式碼：請使用一個 print() 輸出兩行以上的自我介紹
print("""大家好！
我是小明，很高興認識大家。
這是我練習 Python 的作業。""")
print("-" * 30)

# 3. 輸出變數
name = "周彗
萍"
# ✏️ 你的程式碼：請使用 print() 將變數 name 的內容輸出出來
print(name)
print("-" * 30)

# 4. 同時輸出文字與變數
# 提示：使用逗號將文字與變數分隔
# ✏️ 你的程式碼：請輸出「姓名：」(必須使用到上面的 name 變數)
print("姓名：", name)


# 5. 跳脫字元應用 (換行)
# 提示：使用 \n 可以在單一字串中進行換行
# ✏️ 你的程式碼：請只使用一個 print()，利用 \n 讓輸出結果分成兩行顯示「第一行」與「第二行」
print("第一行\n第二行")


# 6. 跳脫字元應用 (跳格 Tab)
# 提示：使用 \t 可以產生類似 Tab 鍵的排版空格
# ✏️ 你的程式碼：請使用 \t 輸出一個簡單的表格標題：「座號 姓名 成績」
print("座號\t姓名\t成績")


# 7. f-string 格式化輸出
score = 95
# ✏️ 你的程式碼：請使用 f-string 配合上述變數，輸出「小明的成績是95分」
print(f"{name}的成績是{score}分")


# 8. 不換行 (更改結尾字元)
# 提示：使用 end=" → "
# ✏️ 你的程式碼：請用兩個 print()，但讓最終輸出的結果在同一行變成「第一關 → 第二關」
print("第一關", end=" → ")
print("第二關")


# 9. 設定分隔符號
# 提示：使用 sep="/"
# ✏️ 你的程式碼：請使用一個 print() 印出三個數字 2026, 9, 4，並讓輸出結果為「2026/9/4」
print(2026, 9, 4, sep="/")


# 10. 格式化浮點數 (限制小數點後兩位)
# 提示：使用 f"{變數:.2f}"
pi = 3.1415926
# ✏️ 你的程式碼：請利用 f-string 輸出「圓周率大約是 3.14」
print(f"圓周率大約是 {pi:.2f}")


# 11. 在 print 內直接進行計算並輸出
# 提示：直接在 f-string 的 {} 中放入算式
price = 250
quantity = 3
# ✏️ 你的程式碼：請輸出「單價 250 元買 3 個，總共花費了 750 元」
print(f"單價 {price} 元買 {quantity} 個，總共花費了 {price * quantity} 元")


# 12. 字串對齊排版 (挑戰題)
# 提示：使用 f"{變數:>10}" 可以向右對齊並佔用 10 個字元寬度
item1 = "蘋果"
item2 = "香蕉"
# ✏️ 你的程式碼：請用 f-string 將上述兩個水果名稱「靠右對齊」輸出
print(f"{item1:>10}")
print(f"{item2:>10}")


# ==========================================
```
